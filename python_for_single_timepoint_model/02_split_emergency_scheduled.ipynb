```python
# ライブラリのインポート
import os
import polars as pl
```

```python
# フォルダ設定
data_dir = '../data/'
output_dir = '../output/'

# date
date = '20251206_first24h'
```

```python
# ファイル名
df_development_file_name = f'{date}_df_development_raw.csv'
df_development_emergency_file_name = f'{date}_emergency_df_development_raw.csv'
df_development_scheduled_file_name = f'{date}_scheduled_df_development_raw.csv'
```

```python
# データの読み込み
df_development = pl.read_csv(os.path.join(data_dir, df_development_file_name))
```

```python
# emergency のみ抽出
df_emergency = df_development.filter(
    pl.col("icu_admission_type") == "emergency"
)

# scheduled_surgery のみ抽出
df_scheduled = df_development.filter(
    pl.col("icu_admission_type") == "scheduled_surgery"
)
```

```python
# 保存
df_emergency.write_csv(
    os.path.join(data_dir, df_development_emergency_file_name)
)

df_scheduled.write_csv(
    os.path.join(data_dir, df_development_scheduled_file_name)
)
```
