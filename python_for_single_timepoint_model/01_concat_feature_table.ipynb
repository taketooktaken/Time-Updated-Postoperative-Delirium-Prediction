# Concat Feature Table

### 作業内容
分割して出力したfeature tableを結合する。

```python
# ライブラリのインポート
import polars as pl
import os
```

### Set Config

```python
# date
date = '20260316_emergency_first24h'

# table名の指定
data_dir = f'../data/{date}/'
export_dir = '../data/'

# 全てのファイル名を取得
file_list = os.listdir(data_dir)
```

### Load data

```python
# 各CSVファイルを読み込み、データフレームのリストを作成
df_list = [pl.read_csv(source=data_dir + file, has_header=True, infer_schema_length=100000,
    schema_overrides = 
        {'fio2_ordered': pl.Float64,
         'peep_ordered': pl.Float64,
         'wbc': pl.Float64,
         'hemoglobin': pl.Float64,
         'platelet': pl.Float64,
         'sodium': pl.Float64,
         'creatinine': pl.Float64,
         'potassium': pl.Float64,
         'aspartate_aminotransferase': pl.Float64,
         'blood_urea_nitrogen': pl.Float64,
         'alanine_aminotransferase': pl.Float64,
         'total_bilirubin': pl.Float64,
         'crp': pl.Float64,
         'albumin': pl.Float64,
         'input_amount': pl.Float64,
         'cumsum_in': pl.Float64,
         'cumsum_out': pl.Float64,
         'cumsum_balance': pl.Float64,
         'window_total_output': pl.Float64,
         'standardized_input_amount': pl.Float64,
         'active_ingredient_propofol': pl.Float64,
         'active_ingredient_noradrenaline': pl.Float64,
         'active_ingredient_fentanyl': pl.Float64,
         'active_ingredient_dexmedetomidine': pl.Float64,
         'active_ingredient_dopamine': pl.Float64,
         'active_ingredient_midazolam': pl.Float64,
         'active_ingredient_dobutamine': pl.Float64,
         'active_ingredient_landiolol': pl.Float64,
         'active_ingredient_vasopressin': pl.Float64,
         'active_ingredient_remifentanil': pl.Float64,
         'active_ingredient_thiamylal': pl.Float64,
         'active_ingredient_adrenaline': pl.Float64,
         'active_ingredient_vecuronium': pl.Float64,
         'active_ingredient_rocuronium': pl.Float64,
         'bg_lactate': pl.Float64,
         'bg_glucose': pl.Float64,
         'inspiratory_pressure': pl.Float64}) for file in file_list]

# データフレームを縦方向（行方向）に結合
df_combined = pl.concat(df_list, how='vertical')
```

### Save as CSV

```python
df_combined.write_csv(os.path.join(export_dir, f'{date}_features.csv'))
```
